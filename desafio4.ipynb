```python
# Desafío 4: Verificar múltiplos de varios números

# Solicitar al usuario que ingrese un número
try:
    numero = int(input("Ingrese un número entero: "))
except ValueError:
    print("Entrada inválida. Por favor ingrese un número entero.")
    # Salir del programa si la entrada no es válida
    exit()

# Lista de divisores a verificar
divisores = [2, 3, 5, 7, 9, 10, 11]

# Verificar si el número es múltiplo de cada divisor
for divisor in divisores:
    if numero % divisor == 0:
        print(f"El número {numero} es múltiplo de {divisor}.")
    else:
        print(f"El número {numero} NO es múltiplo de {divisor}.")
```
