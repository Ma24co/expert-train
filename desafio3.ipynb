```python
# convertir temperatura de celsius a fahrenheit y kelvin
def convertir_temperatura(celsius):
  fahrenheit = (celsius * 9 / 5) + 32
  kelvin = (celsius + 273.15)
  return fahrenheit, kelvin

# solicitar al usuario que ingrese la temperatura en celsius

celsius = float(input("ingresar la temperatura en grados celcius:"))

# realizar las conversiones
fahrenheit, kelvin = convertir_temperatura(celsius)

# mostrar los resultados
# Se corrigió la indentación de estas líneas
print(f"{celsius}°C equivale a {fahrenheit:.2f}°F")
print(f"{celsius}°C equivale a {kelvin:.2f}K")
```
